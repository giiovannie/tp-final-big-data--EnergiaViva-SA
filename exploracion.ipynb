##### `id_medidor`

vamos a analisar si encontramos algo raro 

```python
df["id_medidor"].unique()
```

para este caso vamos a deducir que el formato correcto es ``MED0000`` y le vamos a aplicar a esa columna una exprecion regular que busque algo que no se parezca a esa convencion

```python
df[~df["id_medidor"].str.match(r"^MED\d{4}$")]
```

arriba podemos apreciar que no hay datos que rompan esta conviccion de``MED0000``

```python
df.groupby("id_medidor")["barrio"].nunique().max()
#este analisis me guie de la ia para sacar esta inconsistencia 
```

un medidor físico (MED0099) debería pertenecer siempre al mismo barrio y al mismo tipo_cliente por lo que hay medidores viajando entre barrios 

con esto concluimos la exploracion de esta columna y podemos decir que las duplicaciones creo que se debe a que barrio tiene la misma categoria pero con diferencia de espaciado e iniciales , por eso crea esa duplicacion anomala
